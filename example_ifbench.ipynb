```python
trainset = trainset[:30]
valset = valset[:60]
```

```python
import gepa

seed_prompt = {
    "system_prompt": "Respond to the query"
}
task_lm = "openai/gpt-5-nano"
reflection_lm = "openai/gpt-5"
```

```python
from gepa.adapters.ifbench_adapter.ifbench_adapter import IFBenchAdapter
adapter = IFBenchAdapter(model=task_lm, max_litellm_workers=16)

# Let's run GEPA optimization process.
gepa_result = gepa.optimize(
    seed_candidate=seed_prompt,
    trainset=trainset, valset=valset,
    # task_lm=task_lm, # <-- This is the model being optimized
    max_metric_calls=450, # <-- Set a budget
    reflection_lm=reflection_lm, # <-- Use a strong model to reflect on mistakes and propose better prompts
    adapter=adapter,
)

print("GEPA Optimized Prompt:", gepa_result.best_candidate['system_prompt'])
```
